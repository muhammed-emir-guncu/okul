## temel giriş

```python
class Adam:
    # sınıf yapıcı ve özellikleri
    def __init__(self, ad:str, soyad:str, yas:int) -> None:
        # self burda nesneyi refere eder 
        # self yerine istediğini yazabilirsin 
        # ancak genel kullanımda self yazılır 
        # yanı selfin belli bir özelliği yok
        self.ad = ad
        self.soyad = soyad
        self.yas = yas
        self.canli = True
    # sınıfın metodu
    def tanit(self) -> None:
        print(self.ad ,self.soyad, self.yas, "canlı" if self.canli else "ölü")
    def geber(self) -> None:
        if self.canli:
            self.canli = False
    def dirilt(self) -> None:
        if self.canli == False:
            self.canli = True
    def buyu(self,ek:int=1) -> None:
        self.yas += ek
    def kucult(self,ek:int=1) -> None:
        self.yas -= ek
    
# nesne 
kisi:Adam = Adam("kişi", "soy", 60)
kisi.tanit()
kisi.geber()
kisi.kucult()
kisi.tanit()
kisi.dirilt()
kisi.buyu(2)
kisi.tanit()
```

## miras (kalıtım)

```python
class Konum:
    def __init__(self,X:tuple[int,int]) -> None:
        self.k=[X[0],X[1]]
    def konum_ver(self):
        return (self.k[0],self.k[1])
    def konuma_git(self,u):
        self.k[0]+=u[0]
        self.k[1]+=u[1]

A:Konum=Konum((10,10))
print(A.konum_ver())
A.konuma_git((10,10))
print(A.konum_ver())
    
```

```python
class Konum_2(Konum):
    pass
A=Konum_2((15,15))
A.konum_ver()
A.konuma_git((10,10))
A.konum_ver()
```

```python
class Yer(Konum):
    def konuma_git(self, u):
        self.k[0]=u[0]
        self.k[1]=u[1]
A=Yer((14,14))
A.konum_ver()
A.konuma_git((10,10))
A.konum_ver()
```

```python
class Daire(Konum):
    def __init__(self,X,r):
        super().__init__(X) # super üst sınıfı çağırır
        self.r=r
    def yarichap(self):
        return (3.14)*(self.r**2)
A=Daire((10,10),5)
print(A.yarichap())
print(A.konum_ver())
```

```python
class Daire_1(Konum):
    def __init__(self,X,r):
        self.k=[X[0],X[1]]
        self.r=r
    def yarichap(self):
        return (3.14)*(self.r**2) 
    def tanımlar(self):
        return str(super().konum_ver()) +" "+ str(self.yarichap())
A=Daire_1((10,10),3)
print(A.tanımlar())
```

## kapsülleme

```python
class Banka:
    def __init__(self,sahip,bakiye):
        self.sahip=sahip
        self.__bakiye=bakiye # __ ile başlayan değişkenler dışarıdan erişilemez
    def kyar(self,miktar):
        self.__bakiye+=miktar
    def bilgi_bakiye(self):
        return self.__bakiye # erişilemez değişskene bu şekilde erişebiliriz
A=Banka("emir",100)
A.kyar(100)
print(A.bilgi_bakiye())
print(A.sahip)
# print(A.__bakiye) hata verir
    
```

## dunder metotları (double underscore)

### temel

```python
class Nid:# new init del
    def __new__(cls, *args, **kwargs):
        print("nesne oluşuyor")
        return super().__new__(cls)# anlamadım bunu
    def __init__(self,deger):
        self.deger=deger
        print("nesne oluşturuldu")
    def __del__(self):
        print("nesne siliniyor")
nesne=Nid(10)
del nesne
```

### aritmetik 

```python
class Arit:
    def __init__(self,deger):
        self.deger=deger
    def __add__(self,other):
        return self.deger + other.deger
    def __sub__(self,other):
        return self.deger - other.deger
    def __mul__(self,other):
        return self.deger * other.deger
    def __floordiv__(self,other):
        return self.deger // other.deger
    def __truediv__(self,other):
        return self.deger / other.deger
    def __mod__(self,other):
        return self.deger % other.deger
    def __pow__(self,other):
        return self.deger ** other.deger
A=Arit(20)
B=Arit(30)
print(A+B)
```

### mantık

```python
class Mantuk:
    def __init__(self,deger):
        self.deger=deger
    def __bool__(self):
        return self.deger>=1
    def __eq__(self,other):
        return self.deger==other.deger
    def __ne__(self,other):
        return self.deger!=other.deger
    def __lt__(self,other):
        return self.deger<other.deger
    def __gt__(self,other):
        return self.deger>other.deger
    def __le__(self,other):
        return self.deger<=other.deger
    def __ge__(self,other):
        return self.deger>=other.deger

A=Mantuk(10)
B=Mantuk(20)
print(A==B)
```

### string

```python
class Strng:
    def __init__(self,deger):
        self.deger=deger
    def __str__(self):
        return str(self.deger)
    def __repr__(self):
        return f"Strng({self.deger})"
    def __hash__(self):
        return hash(self.deger)
A=Strng(10)
print(A)
repr(A)

```
